```python
# Finds the first matching element
print(soup.find("a"))
print("\n\n") 
soup.find("p")
```

```python
# Find a tag with specific attributes 
soup.find("p", class_="description")
```

```python
# Finds all matching elements:
soup.find_all("p")
```

```python
# Select elements using CSS selectors.
soup.select_one("p.description")
```

```python
# Handling Missing Elements Safely
# Always check if an element exists before accessing it:
title_tag = soup.find("h1")
if title_tag:
    print(title_tag.text)
else:
    print("Title not found")
```

```python
item = []
for i in range(1,21): 
    with open(f"html/Page{i}.html") as f:
        content = f.read()
    soup = BeautifulSoup(content, "html.parser")
    articles = soup.select("article.product_pod")
    for article in articles:
        title = article.find("h3").find("a")['title']
        price = article.select_one("p.price_color").text.split('£')[1]
        rating_ele = article.select_one("p.star-rating")
        rating = rating_ele['class'][1]
        item.append([title, price, rating])
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(item, columns=['Book Title', 'Price', 'Rating'])
```
