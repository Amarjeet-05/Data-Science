```python
import pandas as pd 
import requests 
from bs4 import BeautifulSoup
```

```python
url = "https://www.ambitionbox.com/it-services-and-consulting-companies-in-india"
```

```python
headers = {"User-Agent" : "Mozilla/5.0"}
```

```python
response = requests.get(url, headers=headers)
```

```python
response
```

```python
soup = BeautifulSoup(response.content, "html.parser")
```

```python
# print(soup.prettify())
```

```python
companies = soup.find_all('div', class_ = 'LocCollections__companyName')
company = []
for comp in companies: 
    company.append(comp.text.strip())
```

```python
company
```

```python
ratings = soup.find_all('div', class_ = "rating_text rating_text--sm")
rating = []
for rate in ratings: 
    rating.append(float(rate.text.strip()))
```

```python
reviews = soup.find_all('span', class_ = 'LocCollections__companyRatingCount')
review = [] 
for rev in reviews: 
    review.append(int(rev.text.strip().split()[0].replace('(', '').strip()))
    
```

```python
review
```

```python
rating
```

```python
logos_html = soup.find_all('a', class_ = 'LocCollections__card')
```

```python
logos = [logo.find('img')['src'] for logo in logos_html]
```

```python
logos
```

```python
data = [] 
for com, rev, rat, log in zip(company, review, rating, logos): 
    data.append({"company" : com, 
                "rating" : rat,
                 "review" : rev,
                 "logo" : log
                })
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
image_paths = [] 

for i, url in enumerate(logos): 
    img_data = requests.get(url).content
    path = f"logos/logo{i}.png"

    with open(path, 'wb') as f:
        f.write(img_data)

    image_paths.append(path)
# df['logo_path'] = image_paths
    df['image'] = img_data
```
