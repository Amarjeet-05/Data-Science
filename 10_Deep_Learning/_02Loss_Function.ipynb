```python
from sklearn.datasets import make_classification
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
x, y = make_classification(n_samples=100, n_features=2, n_classes=2, n_informative=1, 
                           n_redundant=0, n_clusters_per_class=1, random_state=41,
                           hypercube=False, class_sep=15)
```

```python
y = np.where(y == 0, -1, 1)  #we are replacing 0 with -1 because n_classes gives 0 and 1
```

```python
plt.scatter(x[:,0], x[:,1], c=y, s=100)
```

```python
def perceptrone(x,y): 
    w1 = w2 = b = 1
    lr = 0.1 

    for j in range(1000):
        for i in range(0, len(x)): 

            z = w1 * x[i][0] + w2*x[i][1] + b 

            if z*y[i] < 0:   #here it represent this situation if max(0,-yif(xi)) if z*yi<0 -> max(0,+ve)[for incorrect prediction]. if z*yi>0 -> max(0,-ve)[for correct ]
                w1 = w1 + lr*y[i]*x[i][0] 
                w2 = w2 + lr*y[i]*x[i][1] 
                b = b + lr*y[i] 
    return w1, w2, b
    
```

```python
w1,w2,b = perceptrone(x,y)
```

```python
print("w1 : ", w1) 
print("w2 : ", w2) 
print("b : ", b)
```

```python
m = -(w1/w2)
c = -(b/w2)


x_input = np.linspace(-3,3,100)
y_input = m*x_input + c

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(x[:,0],x[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
