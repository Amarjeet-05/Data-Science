```python
import matplotlib.pyplot as plt 
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.plot(years, runs)
 
```

```python
plt.plot(years, runs)
plt.xlabel("Year")
plt.ylabel("Runs")
plt.title("Sachin Tendulkar's yearly runs")
```

```python
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]

plt.plot(years, kohli, label="Kohli")
plt.plot(years, sehwag, label="Sehwag")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Kohli and Sehwag Runs Comparision")
plt.legend()
# plt.legend() showing the which line color represents which data(like kohli represent as blue line and sehwag as orange)
```

# Format string 
we can use format string to plot graph
it can used like fmt = '[marker][line][color]' or [color][marker][line]
we can check format string on matplotlib documentation to know which symbol or mark represent color, line, and marker
format string in this way is not good practice

character           description
'-'                 solid line style
'--'                dashed line style
'-.'                dash-dot line style
':'                 dotted line style

```python
plt.plot(years, kohli, 'ro--', label="Kohli") #ro-- means r= red color, o=marker, -- = line
plt.plot(years, sehwag, 'g^:', label="Sehwag")#g^: mean g=green color, ^=marker, :=doted line
plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend()
```

```python
# format string are not good practice because it is not readable we required check documentation 
# we can use it like:
plt.plot(years, kohli, color= 'red', linestyle='--', marker='o', label='Kohli')
plt.plot(years, sehwag, color= 'blue', linestyle=':', marker='d', label= 'Sehwag')
plt.plot(years, runs, color='brown', linestyle='-.', marker='x', label= 'Tendulkar')
plt.xlabel('Years')
plt.ylabel('Runs')
plt.legend()
```

```python
# we can also change the width of line apply grid  
plt.plot(years, kohli, color= 'red', linestyle='--', marker='o', label='Kohli', linewidth=3)
plt.plot(years, sehwag, color= 'blue', linestyle=':', marker='d', label= 'Sehwag', linewidth=4)
plt.plot(years, runs, color='brown', linestyle='-.', marker='x', label= 'Tendulkar')
plt.xlabel('Years')
plt.ylabel('Runs')
plt.legend()
plt.grid(True)
```
