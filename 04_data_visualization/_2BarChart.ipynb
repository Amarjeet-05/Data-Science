```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
```

```python
# In bar() → x = categories, y = values

plt.bar(years, runs)
plt.xlabel("Yeas")
plt.ylabel("Runs")
```

```python
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
```

```python
x = np.arange(len(years))
width = 0.25
# here we are using the arange function because we have 10 years data so we run arange function to the length of year(10)
# so now instead of plotting on year we are plotting of chart in no.[0,1,2,...9]
# we do this because we want to plot all the players run in a single graph for comparision
# we use width because by default the bar width is 0.8 so as we required to plot three bar in a single plot we minimize the size 
```

```python
plt.bar(x-width, sachin, label="sachin", width=width)
plt.bar(x, sehwag, label="sehwag", width=width)
plt.bar(x+width, kohli, label="kohli", width=width)
plt.xlabel("years")
plt.ylabel("runs")
plt.legend()
# we are using x-width, x, x+width that means it shows sachin's runs at -1.75(0- 0.25), and sehwags at 1, and kohlis at 1.25(1+0.25) location.
# so in this graph we can see the years are 0,1,2,...9
# if we want to show year we use xtick().
```

```python
plt.bar(x-width, sachin, width=width, label = "Sachin")
plt.bar(x, sehwag, width=width, label="Sehwag")
plt.bar(x+width, kohli, width=width, label="Kohli")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend()
plt.xticks(x, years) #now this function helps us to show years
plt.show()

# Why Use xticks()?
# By default, plt.bar() uses numeric x-values (0, 1, 2, ...).
# We use plt.xticks() to set the correct category labels like years or names.

```

```python
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [500+700+1100+1500+1800, 0+200+900+1400+1600, 0+0+500+800+1100, 300+600+800+1100+900]
```

```python
# horizontal bar 
plt.barh(players, runs_5yrs) #in which we first declare ylabel and the x label
plt.xlabel("last 5 years runs")
plt.ylabel("Player's name")

# Why Switch x and y?
# In bar() → x = categories, y = values
# In barh() → y = categories, x = values
```

```python
# Adding Value Labels with plt.text()
# You can use plt.text() to display values on top of bars, making your chart easier to read.

plt.bar(players, runs_5yrs)

for i in range(len(players)):
    plt.text(i, runs_5yrs[i]+40, str(runs_5yrs[i]), ha = "center") 
    #in this function first we declare player(i) and then where we want to show value (run_5yrs[i]+30)
    # means run of player i + 30 here the values shows of the players runs and then we show what we want that is run (run_5yrs[i])
    # and ha="center" means horizontal align 

plt.xlabel("Players")
plt.ylabel("Runs")
```

```python
Products = ["Laptop", "Mobile", "Tablet", "Watch"]
Sales= [250, 400, 150, 200]
```

```python
x = range(len(Products))
plt.bar(x, Sales)

for i in range(len(Products)): 
    plt.text(i, Sales[i] + 3, str(Sales[i]), ha="center")

plt.xlabel("Products")
plt.ylabel("Sales")
plt.xticks(x, Products)
```
