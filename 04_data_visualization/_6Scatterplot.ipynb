```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90]
```

```python
plt.scatter(study_hours, exam_scores)
```

```python
# Size of points based on score (bigger score -> bigger point) 
size = [score *2 for score in exam_scores]
color = ["red" if marks > 60 else "green" for marks in exam_scores]
plt.scatter(study_hours, exam_scores, s=size, c=color)
plt.grid(True, linestyle="--", alpha=0.7)
```

```python
# Also works with Numpy Arrays (used when we use colormap ) we can use different type of cmap color scheme by checking in its documentaion 
score_arr = np.array(exam_scores)
plt.scatter(study_hours, exam_scores, c=score_arr, cmap='viridis')
plt.colorbar(label="Scores")
```
