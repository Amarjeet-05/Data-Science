```python
import matplotlib.pyplot as plt
```

```python
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
```

```python
plt.pie(runs, labels=players)
plt.title("Indian player's runs")
```

```python
# Custom Colors
color = ['#ff9999','#66b3ff','#99ff99','#ffcc99'] #we can use he color code in matplotlib 
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
```

```python
plt.pie(runs, labels=players, colors=color) 
plt.title("Player's Runs")
```

```python
explode = [0.1, 0, 0, 0]  # Only highlight Sachin's slice
plt.pie(runs, labels=players, colors=color, shadow=True, explode=explode) 
plt.title("Player's Runs")
# explode function will sliced the specific slice
```

```python
explode = [0.1, 0.4, 0, 0]  
plt.pie(runs, labels=players, colors=color, shadow=True, explode=explode) 
plt.title("Player's Runs")
```

```python
# we can use wedgeprops it have various function which we can find from wedge documentation of matplotlib
plt.pie(runs, labels=players, wedgeprops={'edgecolor' : 'black', 'linestyle' : '--', 'linewidth':2})
```

```python
# autopct='%1.1f%%' it will percentge in pie chart, .1f%% means how many decimal point will show

plt.pie(runs, labels=players, colors=color, explode=explode, shadow=True, autopct='%1.1f%%')
```

```python
# startangle rotate the chart
explode2 = [0, 0.4, 0, 0]
plt.pie(runs, labels=players, colors=color, shadow=True, autopct='%1.2f%%', startangle=100, explode=explode2)
```
