```python
import matplotlib.pyplot as plt 
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y1 = [i * 2 for i in x]
y2 = [i ** 2 for i in x]
# subplot are using if we want to show graphs side by side 
```

```python
plt.subplot(1,2,1) #subplot(row,column,plotno.)
plt.plot(x,y1)
plt.xlabel("Number")
plt.ylabel("Multiply by 2")
plt.title("x * 2")

plt.subplot(1,2,2)
plt.scatter(x,y2)
plt.xlabel("Number")
plt.ylabel("Square")
plt.title("x ** 2")
plt.tight_layout() #if we our graphs are collaps then we used it
```

```python
y3 = [i ** 0.5 for i in x]
y4 = [10 - i for i in x]
```

```python
# subplots()
fig, axs = plt.subplots(2,2, figsize=(10,6))
axs[0,0].plot(x, y1)
axs[0,0].set_title("x * 2")

axs[0,1].plot(x,y2 )
axs[0,1].set_title("x ** 2")

axs[1,0].plot(x, y3)
axs[1,0].set_title("x ** 0.5")

axs[1,1].plot(x,y4)
axs[1,1].set_title("10-x")

fig.tight_layout()
fig.suptitle("Comparison", fontsize=14)
# fig.savefig("figure.png") for saving as image
```

```python
# using loop
fig, axs = plt.subplots(2,2)
ys = [y1, y2, y3, y4]
titles = ['x * 2', 'x squared', 'sqrt(x)', '10 - x']

for i in range(2):
    for j in range(2):
        idx = i * 2 + j 
        axs[i,j].plot(x, ys[idx])
        axs[i,j].set_title(titles[idx])
fig.tight_layout()
```
