```python
import matplotlib.pyplot as plt 
```

```python
days = [1,2,3,4,5,6,7]
college = [5,5,6,6,5,5,0]
sleep = [7,7,8,8,9,9,10]
gaming = [2,2,1,1,3,4,2]
study = [2,2,2,2,3,2,4]
labels = ['Sleep', 'College', 'Study', 'Gaming']
color = ['green', 'red', 'pink', 'yellow']
```

```python
plt.figure(figsize=(6,6))
plt.stackplot(days, sleep, college, study, gaming, labels=labels, colors=color, alpha=0.7)
plt.xlabel("Days")
plt.ylabel("Hours")
plt.legend(loc="upper left") #loc="upper left" showing legends on upper left
plt.grid(True)
plt.title("Time spent on various activities")
```
