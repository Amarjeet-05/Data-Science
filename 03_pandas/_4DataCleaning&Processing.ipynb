# Type conversion

```python
df['Age'] = df['Age'].astype('int64')
# if any value in age column are null/Nan then we can't do type conversiion we need to drop null or fill null after that we can do type conversion. whenever it will gives eIOError
```

```python
# df['Join DateDate'] = pd.to_datetime(df['Join Date'])
# used to convert date time format
pd.to_datetime(df['Join Date'], format='%d-%m-%Y')
```

# some important function

```python
df['Age Group'] = df['Age'].apply(lambda x: 'Adult' if x >= 25 else 'Minor')
```

```python
df
```

```python
gender = {'M' : 'Male', 'F' : 'Female'}
df['Gender'] = df['Gender'].map(gender)
```

```python
df['City'] = df['City'].replace({'Delhi' : 'New Delhi', 'New York' : 'Chandigarh'})
```
