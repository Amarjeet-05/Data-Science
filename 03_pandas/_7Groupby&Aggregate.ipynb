```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  
```

```python
df
```

```python
df.groupby("Department")['Salary'].mean()
```

```python
df.groupby('Department')['Team'].count()
```

```python
df.groupby("Team")['Salary'].sum()
```

```python
df.groupby("Team")['Salary'].min()
```

```python
df.groupby("Department")['Salary'].max()
```

```python
# .aggregate and .agg functions are same 
df.groupby("Department")[['Salary', 'Age']].agg(['mean', 'min', 'max'])
```

```python
# we can also change column name 
df.groupby("Department")['Salary'].aggregate(avg_mean='mean')
```

```python
df.groupby("Team").agg({
    "Salary": "mean",
    "Age": "max"
})
```

```python
df['Mean_avg'] = df.groupby('Team')['Salary'].transform("mean")
```

```python
df.groupby("Department").filter(lambda x: x['Salary'].mean() > 85)
```
