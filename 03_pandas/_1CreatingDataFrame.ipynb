```python
import numpy as np 
import pandas as pd
```

```python
# creating a data frame though python list
lis = [['Aman', 89], ['Anurag', 67], ['Jishan', 70], ['Abhay', 69]]
```

```python
df = pd.DataFrame(lis, columns=['Name', 'Marks'])
df
```

```python
# data frame using dictionary
dic = {'Name' : ['Aman', 'Golu', 'Jishan', 'Yogender'],
      'KD' : [3.52, 2.54, 4.16, 5.37]}
```

```python
df = pd.DataFrame(dic)
df
```
