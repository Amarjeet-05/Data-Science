```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
```

```python
employees
```

```python
departments
```

```python
pd.merge(employees, departments, on="DeptID") 
# inner join by default 
# it merge the data which is common on both. in it we are merging the data by DeptID so it give DeptID that is common on both
```

```python
pd.merge(employees, departments, on="DeptID", how="left")
# here in left we place employees and in right we place the departments we want on deptid which is present in employees(left)
```

```python
pd.merge(employees,departments,on="DeptID", how="right")
# here we want values from departments which is placed on right so we use how=right
```

```python
pd.merge(employees,departments,on="DeptID", how="outer")
# it merges all the data
```
