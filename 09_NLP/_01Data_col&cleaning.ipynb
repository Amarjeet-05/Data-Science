# cleaning data
there are some steps which use to clean data and prepare it of model

steps:-

1. lowercasing - converting all text in lower case because model will differentiate
                  "The" and "the"

2. removing punctuation - Removing punctuation (!,@#$ etc.) because they don't add
                         meaning in ML context. But if we are using deep learning
                         then we don't remove puntuation.

3. remove numbers - removing digits from the text('i have 5 phones' -> 'i have phones')

4. remove urls and html tags

5. remove emojis and special characters - images uses unicode values(utf) and we can
                             eliminate by considering ASCII value of characters only

6. remove stopwords - one of the most important step removing repitative word which
                      don't carry any meaning eg. is, are,the, have
(chandigarh is the city beautiful -> chandigarh city beautiful)

```python
# now we can easily convert the emotion into numeric value because ml model is interpret only with numbers 
unique_emo = df['emotion'].unique()
emotions = {}
i = 0
for emotion in unique_emo: 
    emotions[emotion] = i 
    i += 1

df['emotion'] = df['emotion'].map(emotions)
```

```python
df
```

```python
# now for text content cour first step is to lowercase the text 
# as in our dataset there is all data already in lowercase but we can do lowercasing
# to understanding concepts

df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
df
```

```python
# now the next step is to removing the punctuations 
# so for removing the punctuation we are using string library 

import string 

def rem_punt(txt):
    return txt.translate(str.maketrans("","",string.punctuation))

# string.punctuation: This is a built-in Python constant that contains a string of all
# standard ASCII punctuation symbols (e.g., ! " # \$ % & ' ( ) * + , - )
# .str.maketrans("", "", string.punctuation): This function creates a translation 
# dictionary. By providing three arguments, you instruct Python to map the characters
# in the third string to None. This tells Python to delete every punctuation mark found
# in the text.text.translate(...): This method applies the translation map to your 
# string, returning a brand new string with all of the mapped characters
# (the punctuation marks) completely deleted.


df['text'] = df['text'].apply(rem_punt)
```

```python
df
```

```python
# now removing numbers 

def rem_no(txt): 
    new = ""
    for i in txt:
        if not i.isdigit(): 
            new = new + i 
    return new 


df['text'] = df['text'].apply(rem_no)
```

```python
df
```

```python
# removing emoji 
def rem_emo(txt): 
    new = ""
    for i in txt: 
        if i.isascii():
            new = new + i 
    return new 

df['text'] = df['text'].apply(rem_emo)
```

```python
df
```

```python
# removing stopwords 
# for removing stopwords we are using nltk library

import nltk
from nltk.corpus import stopwords # corpus means collection of words
from nltk.tokenize import word_tokenize

# each individual word refers as token 
# and the process of splitting the sentence into tokens called tokenization
# we can also use .split() for splitting the words 
# eg = ['hi', 'i', 'am', 'aman']

nltk.download('punkt') # use to split the text into tokens
nltk.download('punkt_tab')
nltk.download('stopwords') 
```

```python
# now lets see stopwords 
len(set(stopwords.words('english')))
# there is 198 stopwords in english language we can see them by removing len() 
```

```python
# lets remove the stopwords and separate the tokens 
stop_word = set(stopwords.words('english'))

def tokenization(txt): 
    texts = word_tokenize(txt)
    tokens = []

    for i in texts: 
        if i not in stop_word: 
            tokens.append(i)
    return " ".join(tokens) #we are using .join() because it convert that list into string

df['text'] = df['text'].apply(tokenization)
```
