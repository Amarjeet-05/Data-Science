```python
import pandas as pd 
import numpy as np 
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
# train_test_split
# generally we are using train_test_split for regression problem. It works well for large datasets
# Can give biased results and Doesn’t handle imbalance well


# StratifiedShuffleSplit
# Only for classification
# Multiple splits → more reliable evaluation
# Works well for imbalanced data
# Computationally more expensive

# When to Use

# Data is imbalanced
# You want multiple random splits
# Dataset is not too large


# there are multiple splitting methods with their own pros and cons 

# Stratified K-Fold → BEST for classification
# Why it’s best:
# Maintains class balance
# Uses full dataset efficiently
# Gives stable and reliable results

# This is the most preferred method in real ML projects



# K-Fold → BEST for regression
# Why:
# No class distribution needed
# Reduces randomness
# Good performance estimate
```

```python
df = pd.read_csv('housing.csv')
```

```python
df
```

```python
# in this dataset, median income is a strong predictor of house prices. Instead of 
# randomly sampling, we can create strata based on income levels (e.g., binning median
# income into categories) and ensure the test set maintains the same distribution of 
# income levels as the full dataset.
```

```python
df['income_cat'] = pd.cut(df['median_income'], bins = [0, 1.5, 3.0, 4.5, 6.0, np.inf], labels=[1,2,3,4,5])
# we are taking bins just like labels example like if salary lies between 0-1.5 then the label is 1,
# if lies between 1.5-3.0 label is 2, lies between 6.0 to infinite(more than 6.0) label is 5.
```

```python
import matplotlib.pyplot as plt
```

```python
df['income_cat'].value_counts().sort_index().plot.bar(rot = 0, grid=True)
plt.title("income category distribution")
plt.xlabel("income category")
plt.ylabel("number of instance")

# as we can clearly see the most number of income category is 3 which can we say as the mean
```

```python
strat = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
```

```python
for train_idx, test_idx in strat.split(df,df['income_cat']): 
    x_train = df.loc[train_idx]
    x_test = df.loc[test_idx]
```
