```python
import pandas as pd 
import numpy as np 
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("housing.csv")
```

```python
df
```

```python
x = df.drop('median_house_value', axis=1) 
y = df['median_house_value']
```

```python
x.drop('ocean_proximity', axis=1, inplace=True)
```

```python
# now i removed the categorical data for easy understanding of the pipeline 
```

```python
from sklearn.pipeline import Pipeline
```

```python
pipe = Pipeline([
    ('imputation', SimpleImputer(strategy='median')),
    ('standardization', StandardScaler()),
])

# here we are impute and then standardize the values. 
# it creates a chain of work like first it do imputaion and then it do standardization
# and we can name the process like i wrote 'imputation' and 'standardization'. if we not want to na,e them 
# then we can use make_pipeline function : 

# from sklearn.pipeline import make_pipeline
# make = make_pipeline(SimpleImputer(strategy='meadian'), StandardScaler())
```

```python
# How It Works

# The pipeline takes a list of steps as (name, transformer) pairs.
# Names must be unique and should not contain double underscores __.
# All intermediate steps must be transformers (i.e., must implement fit_transform()).
# The final step can be either a transformer or a predictor.
```

```python
pipe = pipe.fit_transform(x)
```

```python
pipe = pd.DataFrame(pipe, columns=x.columns, index=x.index)
```
