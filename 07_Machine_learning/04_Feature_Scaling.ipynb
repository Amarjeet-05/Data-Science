```python
# Feature scaling is a crucial machine learning preprocessing step that transforms 
# numerical features to a common range or distribution, preventing variables with larger
# magnitudes from dominating models. It ensures equal contribution from all features, 
# enabling faster convergence in algorithms like gradient descent and better performance 
# in distance-based models like KNN, SVM, and Neural Networks.



# feature sclaing always apllies on training data(features) not on label(output) 
```

```python
import pandas as pd 
import numpy as np 
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv("housing.csv")
```

```python
df
```

```python
x = df.drop('median_house_value', axis=1) 
y = df[['median_house_value']]
```

```python
set(x['ocean_proximity'])
```

```python
encode = OneHotEncoder()
encoded = encode.fit_transform(x[['ocean_proximity']])
```

```python
encoded.toarray()
```

```python
encoded = pd.DataFrame(encoded.toarray(), columns=['<1H OCEAN', 'INLAND', 'ISLAND', 'NEAR BAY', 'NEAR OCEAN'], index=x.index)
```

```python
x = x.join(encoded).drop('ocean_proximity',axis=1)
```

```python
# now we have two methods for feature scaling
# 1. MinMaxScaler 
# 2. StandardScaler
```

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
```

```python
#1. MinMaxScaler - it is very simple feature scaling technique in which we give the scale
# like give values between 0-1, 0-10 , -1 - 1, 100-3897, etc. 
# formula = (x-min)/(max-min) 

# but in this approach if there is any outlier then outlier may impact the scale which is not good 
# ex if we have student weight if there is outlier like 975 it will become the max and
# aside from it we have weights like in between 40-100 so the outlier may impact the scale 

# Sensitive to outliers — extreme values can distort the scale.
```

```python
m = MinMaxScaler(feature_range=(-1,1))
minmax = m.fit_transform(x)
```

```python
minmax
```

```python
minmax = pd.DataFrame(minmax, columns=x.columns, index=x.index)
```

```python
minmax
```

```python
# 2. StandardScaler (Z-score)
# This method centers the data around 0 and scales it based on standard deviation.
# it can handle outliers because it work with mean and the standard deviation of the feature
```

```python
stand = StandardScaler()
```

```python
stand = stand.fit_transform(x)
```

```python
stand
```

```python
stand = pd.DataFrame(stand, columns=x.columns, index=x.index)
```

```python
stand
```
