```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.cluster import KMeans, DBSCAN
from sklearn.datasets import make_moons
```

```python
x, y_true = make_moons(n_samples=500, noise=0.05, random_state=45)

df = pd.DataFrame(x, columns=['f1', 'f2'])
```

```python
scaler = StandardScaler() 
x_scaled = scaler.fit_transform(df)
```

```python
# kmeans

kmeans = KMeans(n_clusters=2, random_state=45)
k_labels = kmeans.fit_predict(df)

df['k_cluster'] = k_labels
sns.scatterplot(x = df['f1'], y = df['f2'], hue=df['k_cluster'])
```

```python
# dbscan 

dbscan = DBSCAN(eps=0.3, min_samples=6)
db_labels = dbscan.fit_predict(x_scaled)

df['d_cluster'] = db_labels

sns.scatterplot(x=df['f1'], y=df['f2'], hue=df['d_cluster'])
```
