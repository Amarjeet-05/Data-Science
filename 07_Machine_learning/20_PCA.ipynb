```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

PCA reduces the number of features in a dataset while trying to preserve as much important information (variance) as possible.

Why do we need PCA? 

Large datasets can have many features. This can cause:
- More computation time
- More memory usage
- Difficulty in visualizing data
- Redundant/correlated features
- Sometimes poorer ML performance due to the curse of dimensionality

PCA helps by reducing these dimensions.

```python
x, y_true = make_blobs(n_samples=700, n_features=7, random_state=42)

scale = StandardScaler()
x_scaled = scale.fit_transform(x)
```

```python
x_scaled
# there are 7 columns in our dateset and we are going to apply pca on it. We only want to work with 2 columns, so we have to apply pca.
```

```python
pca = PCA(n_components=2) #n_components = 2 refers that we only have to use 3 features
x_pca = pca.fit_transform(x_scaled)
```

```python
x_pca
```

```python
df_pca = pd.DataFrame(x_pca, columns=['pc1', 'pc2'])
df_pca['lable'] = y_true

sns.scatterplot(x=df_pca['pc1'], y=df_pca['pc2'], hue=df_pca['lable'])
```
