```python
# In this section, we will consolidate everything we've done so far into one final script using Scikit-Learn pipelines. This includes:

# Creating a stratified test set
# Handling missing values
# Encoding categorical variables
# Scaling numerical features
# Combining everything using Pipeline and ColumnTransformer
```

```python
import pandas as pd 
import numpy as np 
from sklearn.impute import SimpleImputer
from sklearn.model_selection import StratifiedShuffleSplit
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
# 1 load the dataset
df = pd.read_csv('housing.csv')
```

```python
# 2 creating stratified test based on income
df['income_cat'] = pd.cut(df['median_income'], bins=[0, 1.5, 3.0, 4.5, 6.0, np.inf], labels=[1, 2, 3, 4, 5])

strat = StratifiedShuffleSplit(n_splits=1, random_state=42, test_size=0.2)

for train_idx, test_idx in strat.split(df, df['income_cat']):
    x_train = df.loc[train_idx].drop('income_cat', axis=1)
    x_test = df.loc[test_idx].drop('income_cat', axis=1)

# now we are working on copy of training data beacause it is a good practice to work with copy of data to prevent changing in original data
training =  x_train.copy()
```

```python
# 3 separate data into features and labels 
x = training.drop('median_house_value', axis=1)
y = training['median_house_value']
```

```python
# 4 list the numerical and categorical data 
x_num = x.drop('ocean_proximity', axis=1).columns.to_list()
x_cat = ['ocean_proximity']
```

```python
# Pipelinesl: 

# i- Numerical pipeline 
num_pipe = Pipeline([
    ('imputer', SimpleImputer(strategy='median')), 
    ('stand', StandardScaler())
])

# ii- Categorical pipeline 
cat_pipe = Pipeline([
    ('hotencode', OneHotEncoder(handle_unknown='ignore'))
])


#  iii-full pipeline 
final_pipe = ColumnTransformer([
    ('num', num_pipe, x_num),
    ('cat', cat_pipe, x_cat)
])
```

```python
final_pipe.fit_transform(x)
```
