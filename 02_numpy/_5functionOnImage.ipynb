```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = plt.imread(r"free-nature-images.jpg")
```

```python
# here it read the image as in the form of array
df
```

```python
plt.imshow(df)
# this command is used to place the image in graph
```

```python
# we can also crop the image using the slicing 
# example if we want to only tree then we take y- axis from 220 - 450 and x-axis from 430-750 
plt.imshow(df[220:451, 430:751])

# this will show cropped image according to the graphical representation of the image 
```

```python
# this loop help to show all changes/filter in the image if we subtract values from 0-255 

for i in range(0, 256): 
    plt.imshow(df-i)
    plt.show()        #we are using this to show each image after subtracting the values 
    display(clear=True)
    
```
