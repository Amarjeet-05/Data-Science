```python
import numpy as np
```

```python
arr = np.array([[1,2,3], [4,5,6]])
```

```python
print(arr.sum(axis = 0))
print(arr.sum(axis=1))
print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))
```

```python
print(arr.min(axis = 0))
print(arr.min(axis=1))
print(np.min(arr, axis=0))
print(np.min(arr, axis=1))
```

```python
print(arr.max(axis = 0))
print(arr.max(axis=1))
print(np.max(arr, axis=0))
print(np.max(arr, axis=1))
```

```python
print(arr.mean(axis = 0))
print(arr.mean(axis=1))
print(np.mean(arr, axis=0))
print(np.mean(arr, axis=1))
```

```python
arr = np.array([1,2,3,4,5])
```

```python
np.std(arr) #standard deviation
#or
arr.std()
```

```python
#we can use function via np.functionname(variable) or variable.functionname
```

```python
np.var(arr)    #variance
```

```python
np.argmax(arr) #index of max element
```

```python
arr.argmin() #index of min element
```

```python
arr.prod() #product
```

```python
np.percentile(arr, 50) #this function not follows variable.functionname()
```

```python
np.corrcoef(arr) #correlation coefficient
# #this function not follows variable.functionname()
```

```python
np.unique(arr) #unique elements
```

```python
np.diff(arr) #nth difference
# 2 - 1 = 1
# 3 - 2 = 1
# 4 - 3 = 1
# 5 - 4 = 1
```

```python
np.cumsum(arr) #cummulative sum
# kind of fibonacci series 
# we have 1,2,3,4,5
#  1, 1+2 = 3, 3+3 =6, 6+4=10....... 
```

```python
np.log(arr) #logrithmoc
```

```python
np.exp(arr) #exponential
```
