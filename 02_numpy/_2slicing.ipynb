```python
# we can also reshape an array if we don't know about its length and calculation of the rows and cols
# but we need to defines row or col in a way so it can create a matrices
# not like that (5,3) if we have 20 elements it is not possible to create (5,3) matrices
arr = np.arange(20)
reshape2dcol = arr.reshape(-1, 5) #here -1 is unknown we only write here how many cols we needed and it automatically defines the rows
print(f"Reshaped 2d array column wise : {reshape2dcol}")

reshape2drow = arr.reshape(5,-1)#here we defines the row and it automatically takes the no. of cols 
print(f"Reshaped 2d array row wise : {reshape2drow}")

reshaped3d = arr.reshape(1,5,-1)
print(f"3d array : {reshaped3d}")


# conerting 2d,3d array in flatten in another way:
arr2 = reshaped3d.reshape(-1)#put -1 it convert in into flatten array
print(f"flatten array : {arr2}")
```
